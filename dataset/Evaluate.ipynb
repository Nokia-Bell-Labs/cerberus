### Check Steering Performance

```python
def evaluate(df,split_str = '[/INST]', filter_word_1 = 'private', filter_word_2 = 'sensitive', double_mode=True):
    df['og_cl'] = ''
    df['new_cl'] = ''
    ids = []
    for i in range(df.shape[0]):
        # split generated output to get output portion only
        df.loc[i,'og_cl'] = df.loc[i,'OG_res'].split(split_str)[1]
        df.loc[i,'new_cl'] = df.loc[i,'New_res'].split(split_str)[1]
        # check if response contain the target words
        if filter_word_1 in df.loc[i,'new_cl']:
            ids.append(i)
        elif double_mode and filter_word_2 in df.loc[i,'new_cl']:
            ids.append(i)
    print(len(ids)/df.shape[0])
    return ids

def has_numbers(inputString):
    return any(char.isdigit() for char in inputString)
```

### Check Baseline Performance

```python
coefs = [0.6,0.7,0.8,0.9,1.0,1.1,1.2,1.3,1.4]
for coef in coefs:
    with open('./data/baseline_sent_'+str(coef)+'.json', 'r') as f:
        res = json.load(f)
    ids = []
    for i in range(len(res)):
        r = res[i]
        clean = r['New_res'].split('[/INST]')[1]
        if 'private' in clean:
            ids.append(i)
        elif 'sensitive' in clean:
            ids.append(i)
    print(coef)
    print(len(res),len(ids))
    print(len(ids)/len(res))
```
